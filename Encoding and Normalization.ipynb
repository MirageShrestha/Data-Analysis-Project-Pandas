```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
```

```python
# Load California housing dataset
housing = fetch_california_housing()
df = pd.DataFrame(data=housing.data, columns=housing.feature_names)
df
```

```python
# Add target variable to the DataFrame
df['MedHouseVal'] = housing.target
df.head()
```

```python
# Check for Missing Values
print(df.isnull().sum())
```

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
df
```

```python
# Encoding Categorical Variables
scaler = StandardScaler()
df[['MedInc', 'HouseAge', 'AveRooms', 'AveOccup', 'Latitude', 'Longitude', 'MedHouseVal']] = scaler.fit_transform(
    df[['MedInc', 'HouseAge', 'AveRooms', 'AveOccup', 'Latitude', 'Longitude', 'MedHouseVal']]
)
df.head()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load the Titanic dataset
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

df.head()
```

```python
# Check for Missing Values
print("\nMissing Values Before Handling:")
print(df.isnull().sum())
```

```python
# Handling Missing Values
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
# Dropping the column due to a high percentage of missing values
df.drop(columns=['Cabin'], inplace=True)
```

```python
print(df.isnull().sum())
```

```python
# Step 4: Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
# Encoding for 'Sex' and 'Embarked'
df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)
df.head()
```

```python
# Normalize Numerical Features
scaler = StandardScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
df.head()
```
