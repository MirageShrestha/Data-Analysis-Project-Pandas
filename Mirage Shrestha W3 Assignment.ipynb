```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import pandas as pd

tips = sns.load_dataset('tips')

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day')
plt.title('Total Bill vs Tip by Day')
plt.show()
```

```python
import pandas as pd
df = pd.read_csv('top 120 best-selling mobile phones.csv')
```

```python
df.head()
```

```python
c_name = ['Rank', 'Manufacturers', 'Model', 'Form_factor', 'Smartphone', 'Year', 'Units_sold_million']
df.columns= c_name
df.head(7)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(df, x='Year', y = 'Units_sold_million', hue = 'Manufacturers')
plt.title('Units sold over years for different manufactures')
plt.legend(prop={'size': 5})
plt.show()
```

```python
plt.figure(figsize=(27,11))
sns.barplot(df, x='Manufacturers', y='Units_sold_million', estimator=sum)
plt.title('Comparing the total units sold by each manufacturer')
plt.show()
```

```python
corr_matrix=df[['Rank', 'Year','Units_sold_million']].corr()
sns.heatmap(corr_matrix, annot = True)
plt.title('Correlation between numerical variables')
```

```python
sns.scatterplot(df, x='Year', y='Units_sold_million', hue='Smartphone', style = 'Form_factor')
plt.title('Unit Sold vs Year by Smartphone and Form Factor')
plt.grid(True)
plt.legend(prop={'size': 6})
plt.show()
```
