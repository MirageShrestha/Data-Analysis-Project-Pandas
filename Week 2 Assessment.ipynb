```python
import pandas as pd
```

Load the data from the csv file into the dataframe of pandas.

```python
global_df = pd.read_csv('Global EV Data 2024.csv')
global_df
```

Display the first few rows of the DataFrame to understand its structure

```python
global_df.head()
```

Filtering: Select the rows where the 'year' = 2012

```python
df_filter = global_df[global_df.year==2012]
df_filter.head()
```

Grouping: Group data by 'region' and 'category'

```python
group = df_filter.groupby(['region', 'category'])
group
```

Aggregating: Calculate the sum of 'value' for each group

```python
agg = group['value'].sum().reset_index()
agg
```

```python
agg_sort=agg.sort_values(by='value', ascending = False)
agg_sort
```
